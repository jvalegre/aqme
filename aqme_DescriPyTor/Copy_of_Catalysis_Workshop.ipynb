```python
from rdkit import Chem
from rdkit.Chem import AllChem
import numpy as np
import pandas as pd

def compute_rdkit_dipole(smiles):
    """Compute approximate dipole vector and magnitude from Gasteiger charges."""
    mol = Chem.AddHs(Chem.MolFromSmiles(smiles))
    AllChem.EmbedMolecule(mol)
    AllChem.UFFOptimizeMolecule(mol)
    AllChem.ComputeGasteigerCharges(mol)

    conf = mol.GetConformer()
    coords = np.array([list(conf.GetAtomPosition(i)) for i in range(mol.GetNumAtoms())])
    charges = np.array([float(a.GetProp("_GasteigerCharge")) for a in mol.GetAtoms()])

    # Dipole vector in e·Å (multiply by e·Å→Debye: 4.80320427)
    dipole_vec = np.sum(charges[:, None] * coords, axis=0) * 4.80320427
    dipole_mag = np.linalg.norm(dipole_vec)

    return dipole_vec, dipole_mag

# Example substituted benzaldehydes
molecules = {
    "H": "O=CC1=CC=CC=C1",
    "p-NO2": "O=CC1=CC=C(C=C1)[N+](=O)[O-]",
    "p-OMe": "O=CC1=CC=C(C=C1)OC",
    "p-CF3": "O=CC1=CC=C(C=C1)C(F)(F)F",
    "p-NH2": "O=CC1=CC=C(C=C1)N"
}

df = []
for name, smi in molecules.items():
    vec, mag = compute_rdkit_dipole(smi)
    df.append({"name": name, "μx": vec[0], "μy": vec[1], "μz": vec[2], "μ_total": mag})

dipole_df = pd.DataFrame(df)
print(dipole_df)
```

```python
import py3Dmol

def show_dipole(smiles):
    mol = Chem.AddHs(Chem.MolFromSmiles(smiles))
    AllChem.EmbedMolecule(mol)
    mb = Chem.MolToMolBlock(mol)
    viewer = py3Dmol.view(width=300, height=300)
    viewer.addModel(mb, "mol")
    viewer.setStyle({'stick': {}})
    viewer.zoomTo()
    return viewer

show_dipole("O=CC1=CC=C(C=C1)[N+](=O)[O-]")
```

```python
import plotly.graph_objects as go
from rdkit import Chem
from rdkit.Chem import AllChem
import numpy as np
import math

def compute_rdkit_dipole(smiles):
    """Embed molecule, optimize, compute Gasteiger dipole vector (Debye)."""
    mol = Chem.AddHs(Chem.MolFromSmiles(smiles))
    AllChem.EmbedMolecule(mol, randomSeed=42)
    AllChem.UFFOptimizeMolecule(mol)
    AllChem.ComputeGasteigerCharges(mol)

    conf = mol.GetConformer()
    coords = np.array([list(conf.GetAtomPosition(i)) for i in range(mol.GetNumAtoms())])
    charges = np.array([float(a.GetProp("_GasteigerCharge")) for a in mol.GetAtoms()])
    dipole_vec = np.sum(charges[:, None] * coords, axis=0) * 4.80320427  # e·Å → D
    return coords, dipole_vec, np.linalg.norm(dipole_vec)

def visualize_dipoles_grid(molecules_dict, n_cols=3, size=250):
    """
    Visualize dipole arrows for multiple molecules in grid layout.
    Each subplot shows atoms and dipole vector.
    """
    n_mols = len(molecules_dict)
    n_rows = math.ceil(n_mols / n_cols)
    specs = [[{'type': 'scene'} for _ in range(n_cols)] for _ in range(n_rows)]

    fig = make_subplots(
        rows=n_rows, cols=n_cols,
        specs=specs,
        horizontal_spacing=0.05,
        vertical_spacing=0.05,
        subplot_titles=[f"{k}" for k in molecules_dict.keys()]
    )

    colors = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728", "#9467bd", "#17becf", "#bcbd22"]

    for i, (name, smi) in enumerate(molecules_dict.items()):
        coords, dip_vec, mag = compute_rdkit_dipole(smi)
        row = i // n_cols + 1
        col = i % n_cols + 1

        # plot atoms
        fig.add_trace(go.Scatter3d(
            x=coords[:, 0], y=coords[:, 1], z=coords[:, 2],
            mode="markers",
            marker=dict(size=3, color=colors[i % len(colors)]),
            name=f"{name} (μ={mag:.2f} D)",
            showlegend=False
        ), row=row, col=col)

        # dipole vector (arrow)
        fig.add_trace(go.Cone(
            x=[0], y=[0], z=[0],
            u=[dip_vec[0]], v=[dip_vec[1]], w=[dip_vec[2]],
            sizemode="absolute", sizeref=5,
            anchor="tail", showscale=False,
            colorscale=[[0, colors[i % len(colors)]], [1, colors[i % len(colors)]]],
            name=f"{name} dipole",
            showlegend=False
        ), row=row, col=col)

        # set subplot title with dipole magnitude
        fig.layout.annotations[i].text = f"{name} (μ={mag:.2f} D)"

    # Global layout
    for r in range(1, n_rows + 1):
        for c in range(1, n_cols + 1):
            fig.update_scenes(
                xaxis_title="", yaxis_title="", zaxis_title="",
                aspectmode="cube", dragmode=False,
                row=r, col=c
            )

    fig.update_layout(
        title="Dipole Vectors of Substituted Benzaldehydes",
        width=n_cols * size * 2,
        height=n_rows * size * 2,
        showlegend=False
    )
    fig.show()

# --- Usage ---
from plotly.subplots import make_subplots
import plotly.io as pio
pio.renderers.default = "notebook"  # ensure Colab display

molecules = {
    "H": "O=CC1=CC=CC=C1",
    "p-NO2": "O=CC1=CC=C(C=C1)[N+](=O)[O-]",
    "p-OMe": "O=CC1=CC=C(C=C1)OC",
    "p-CF3": "O=CC1=CC=C(C=C1)C(F)(F)F",
    "p-NH2": "O=CC1=CC=C(C=C1)N"
}

visualize_dipoles_grid(molecules)
```
